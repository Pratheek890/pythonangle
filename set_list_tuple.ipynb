## exception handling

An exception is an error that occurs during the execution of code. 
differet types of error examples:
1) ZeroDivisionError occurs when you try to divide by zero.
2) NameError -- in this case, it means that you tried to use the variable a when it was not defined.
3) IndexError -- in this case, it occured because you tried to access data from a list using an index that does not exist for this list.
4) ValueError -- This error occurs when an inappropriate value is used within the code
5) FileNotFoundError - This exception is encountered when an attempt is made to access a file that does not exist.
