# Exercise1: Simple intrest calculator

```python
#simple intrest = p*t*r/100 p- principle amount t-time r-rate of intrest
p = int(input("Principle amount: "))        #takes on integer as input
year = int(input("Total years: "))            #total number of years
month = int(input("Total months: "))        #total number of months
t = year + (month / 12)                     #takes year and months and transform to whole number  
print("The total time period is: ",t)
r = float(input("Rate of intrest: "))        #float = decimal points takes whole number an ddecimal points
SI = p*t*r/100                                #simple intrest formula
print("Total intrest amount to be paid " , SI)   #intrest amount
print("Total return amount : ",round(p+SI , 3))   #total amount to be paid to return amount 
```

# **Exercise 2 : fibonacci Series**

#### **The keyword argument end = "" can be used to avoid the newline after the output, or end the output with a different string**



```python
n = int(input("Enter the length of series"))

a ,b = 0,1
while a < n:
    print(a , end=",")
    a,b=b,a + b 
```
