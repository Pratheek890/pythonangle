## break and continue

```python
- break - breaks the inner part of for and while loop 
- continue - continues with the next iteration of the loop 
- else clause in loop - In a for or while loop the break statement may be paired with an else clause. 
    If the loop finishes without executing the break,the else clause executes.
- pass - does nothing just like you just passes to next statement
        
```

```python
#Excerice3 : build a calculator using loops and match case and function 
#in the main function we asked user to input the n number of values after each iternation y/n is taken as input so we get to know weather user wants to continue or not 

def main():
    result = float(input("Enter first number: "))            #the first number taken as input
    while True:                                              #while loop executes when contition is true here everything is true just like deluision
        op = input("Enter operation (+ - * /): ")            #entering the operation + - * / 
        b = float(input("Enter next number: "))              #entering the number b 
        result = calculation(result, b, op)                  #calling second function as calcultion the answer of 2nd calculation is again comes as imput to previous variable result(a/first number)
        
        print("Current Total =", result)
        choice = input("Continue? (y/n): ")                   #ask if user wnat to continue or not 
        if choice.lower() == "n":                             # if user yner Y/y or N/n the it returns the an
            print("Final Answer =", result)                   #if user enter N/n it display the total result 
            break                                             #break statement - breaks the loop 
def calculation(a,b,op):
#op is the operation where + - * / is entered based on op it gives output
    match op:
        case "+":
           return a + b
        case "-":
            return a - b
        case "*":
            return a * b
        case "/":
            if b!=0:
                return a / b
            else:
                print("Invalid operator")
                return a

main()                                                         #calling main function 
                  
```
